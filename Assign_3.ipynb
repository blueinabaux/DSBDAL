```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('nba.csv')
df
```

```python
df.isnull().sum()
```

```python
df_new = df.copy()

# Fill 'College' NaNs with 'Not in a college'
df_new['College'] = df_new['College'].fillna('Not in a college')

# Fill 'Salary' NaNs with the mean salary
df_new['Salary'] = df_new['Salary'].fillna(df_new['Salary'].mean())

# Drop any other NaNs
df_new.dropna(inplace=True)

# Check if any NaNs are left
print(df_new.isnull().sum())
```

```python
age_col = ['0-10','11-15','16-20','21-25','26-30','30-40']

age_series = df_new['Age']
for i in range(len(age_series)):
    if age_series[i] < 10: age_series[i] = age_col[0]
    elif age_series[i] < 15: age_series[i] = age_col[1]
    elif age_series[i] < 20: age_series[i] = age_col[2]
    elif age_series[i] < 25: age_series[i] = age_col[3]
    elif age_series[i] < 30: age_series[i] = age_col[4]
    else: age_series[i] = age_col[5]
    
```

```python
age_series.head(20)
```

```python
df_new['Age'] = df_new['Age'].astype(object)
df1 = df_new.groupby('Age')
df1.head()
# df1.describe()
```

```python
columns = ["Salary"]

for ageGroup, group in df1:
    print(f"Statistics for {ageGroup}:\n")

    for col in columns:
        val = group[col]

        variance = sum((val - val.mean())**2) / len(val)
        deviation = variance**0.5

        print(f"Column: {col}")
        print(f"Variance = {variance}")
        print(f"Standard Deviation = {deviation}")
        print(f"Mean = {val.mean()}")
        print(f"Median = {val.median()}")
        print(f"Min = {val.min()}")
        print(f"Max = {val.max()}")
        print("\n")
```

```python
dfg = df_iris.groupby('Species')
dfg.describe()
```

```python
columns = ['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']

for species, group in dfg:
    print(f"Statistics of {species}:\n")

    for col in columns:
        val = group[col]

        variance = sum((val - val.mean())**2) / (len(val))
        deviation = variance**0.5

        print(f"Column: {col}")
        print(f"Variance = {variance}")
        print(f"Standard Deviation = {deviation}")
        print(f"Mean = {val.mean()}")
        print(f"Median = {val.median()}")
        print(f"Min = {val.min()}")
        print(f"Max = {val.max()}")
        print("\n")
```
