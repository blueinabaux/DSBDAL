```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("Iris.csv")
df
```

```python
df = df.rename(columns = {"SepalLengthCm":"sepal_length", "SepalWidthCm":"sepal_width", "PetalLengthCm":"petal_length", "PetalWidthCm":"petal_width"})
df.head()
```

```python
df.isnull().sum()
```

```python
sns.histplot(data=df, x="sepal_length", kde=True)
```

```python
sns.histplot(data=df, x="sepal_width", kde=True)
```

```python
sns.histplot(data=df, x="petal_length", kde=True)
```

```python
sns.histplot(data=df, x="petal_width", kde=True)
```
