```python
import nltk
```

```python
nltk.download('popular')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
with open("sample.txt", "r") as file:
    dm = file.read()
file.close()

print("File Content:\n", dm)
```

```python
from nltk import sent_tokenize, word_tokenize

sentence_tokens = []
sentence_tokens = sent_tokenize(dm)
print("SENTENCE TOKENS: \n", sentence_tokens)
print("\n")

word_tokens = []
word_tokens = word_tokenize(dm)
print("WORD TOKENS: \n", word_tokens)
```

```python
from nltk.corpus import stopwords

stopwords = stopwords.words('english')
filtered_word_tokens = []

for word in word_tokens:
    if word not in stopwords:
        filtered_word_tokens.append(word)

print("FILTERED WORD TOKENS: \n", filtered_word_tokens)
```

```python
from nltk import pos_tag

pos_tags = []

pos_tags = pos_tag(filtered_word_tokens)

print("POS TAGGING: \n", pos_tags)
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
stemmed_words = []

for word in filtered_word_tokens:
    stemmed_words.append(ps.stem(word))

print("STEMMED WORDS: \n",stemmed_words )
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_words = []

for word in filtered_word_tokens:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print("LEMMATIZED WORDS: \n",lemmatized_words )
```
